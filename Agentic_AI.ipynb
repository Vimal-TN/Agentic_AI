```python
# ============================================================
# MULTI-AGENT LLM MODEL
# Distributed Puzzle Solving
# Architectures:
# 1. Centralized
# 2. Hierarchical
# 3. Peer-to-Peer
# ============================================================

# Install required libraries
!pip install -q -U transformers accelerate

# ============================================================
# IMPORTS
# ============================================================

from transformers import pipeline
import torch
import textwrap

# ============================================================
# LOAD LLM MODEL
# ============================================================

print("Loading LLM model...")

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

llm = pipeline(
    "text-generation",
    model=MODEL_NAME,
    device=0 if torch.cuda.is_available() else -1
)

print("Model loaded successfully!")
print("=" * 70)


# ============================================================
# LLM FUNCTION
# ============================================================

def ask_llm(prompt, max_tokens=180):

    result = llm(
        prompt,
        max_new_tokens=max_tokens,
        do_sample=False,
        return_full_text=False
    )

    return result[0]["generated_text"].strip()


# ============================================================
# AGENT FUNCTION
# ============================================================

def run_agent(agent_name, role, task, puzzle):

    prompt = f"""
You are an AI agent participating in a distributed puzzle-solving system.

Agent Name:
{agent_name}

Your Role:
{role}

Task:
{task}

Puzzle:
{puzzle}

Instructions:
1. Analyze the puzzle carefully.
2. Perform logical reasoning.
3. Give your proposed answer.
4. Explain the important reasoning briefly.
5. Do not discuss programming.
6. Focus only on solving the puzzle.

Answer:
"""

    response = ask_llm(prompt)

    return {
        "agent": agent_name,
        "role": role,
        "response": response
    }


# ============================================================
# GET PUZZLE FROM USER
# ============================================================

print("\n")
print("=" * 70)
print("        MULTI-AGENT DISTRIBUTED PUZZLE SOLVER")
print("=" * 70)

print("\nEnter your own puzzle.")
print("You can enter multiple lines.")
print("Type END when you have finished entering the puzzle.")
print("-" * 70)

lines = []

while True:

    line = input()

    if line.strip().upper() == "END":
        break

    lines.append(line)

puzzle = "\n".join(lines)

print("\n" + "=" * 70)
print("YOUR PUZZLE")
print("=" * 70)
print(puzzle)
print("=" * 70)


# ============================================================
# ARCHITECTURE 1
# CENTRALIZED ARCHITECTURE
# ============================================================

print("\n")
print("#" * 70)
print("ARCHITECTURE 1 : CENTRALIZED")
print("#" * 70)

print("\nCentral Coordinator is assigning the puzzle...")

central_result = run_agent(
    "Central Coordinator",
    "Main coordinator responsible for solving the complete puzzle.",
    """
Solve the puzzle directly.
Analyze all clues and determine the final answer.
Provide a clear logical explanation.
""",
    puzzle
)

print("\n--- CENTRAL COORDINATOR ---")
print(central_result["response"])


# ============================================================
# ARCHITECTURE 2
# HIERARCHICAL ARCHITECTURE
# ============================================================

print("\n")
print("#" * 70)
print("ARCHITECTURE 2 : HIERARCHICAL")
print("#" * 70)


# -----------------------------
# LEAD AGENT
# -----------------------------

print("\n[1] Lead Reasoning Agent")

lead_result = run_agent(
    "Lead Reasoning Agent",
    "Leader of the hierarchical agent group.",
    """
Study the complete puzzle.
Identify the major logical problem.
Develop a possible solution strategy.
Send the important reasoning to the lower-level agents.
""",
    puzzle
)

print(lead_result["response"])


# -----------------------------
# CONSTRAINT AGENT
# -----------------------------

print("\n[2] Constraint Agent")

constraint_result = run_agent(
    "Constraint Agent",
    "Analyzes restrictions, conditions and constraints.",
    """
Identify all important constraints in the puzzle.
Determine which possibilities are allowed or eliminated.
Use logical elimination to find the answer.
""",
    puzzle
)

print(constraint_result["response"])


# -----------------------------
# PATTERN AGENT
# -----------------------------

print("\n[3] Pattern Agent")

pattern_result = run_agent(
    "Pattern Agent",
    "Searches for patterns and relationships.",
    """
Look for numerical, logical, sequential or structural patterns.
Use those patterns to determine the solution.
Explain the important pattern.
""",
    puzzle
)

print(pattern_result["response"])


# -----------------------------
# VALIDATOR AGENT
# -----------------------------

print("\n[4] Validator Agent")

validator_result = run_agent(
    "Validator Agent",
    "Checks the correctness of possible solutions.",
    """
Independently solve the puzzle.
Check possible answers carefully.
Identify errors or contradictions.
Provide the most logically valid answer.
""",
    puzzle
)

print(validator_result["response"])


# ============================================================
# HIERARCHICAL FUSION
# ============================================================

print("\n")
print("-" * 70)
print("HIERARCHICAL AGENT FUSION")
print("-" * 70)

hierarchical_prompt = f"""
You are the Lead Coordinator of a hierarchical multi-agent
puzzle-solving system.

Original Puzzle:
{puzzle}

The following agents analyzed the puzzle.

LEAD REASONING AGENT:
{lead_result["response"]}

CONSTRAINT AGENT:
{constraint_result["response"]}

PATTERN AGENT:
{pattern_result["response"]}

VALIDATOR AGENT:
{validator_result["response"]}

Compare all the responses.

Determine:
1. The correct final answer.
2. The reasoning that supports it.
3. Any disagreement between agents.

Give one final solution.
"""

hierarchical_answer = ask_llm(
    hierarchical_prompt,
    max_tokens=250
)

print("\nHIERARCHICAL FINAL ANSWER:")
print(hierarchical_answer)


# ============================================================
# ARCHITECTURE 3
# PEER-TO-PEER ARCHITECTURE
# ============================================================

print("\n")
print("#" * 70)
print("ARCHITECTURE 3 : PEER-TO-PEER")
print("#" * 70)


# -----------------------------
# PEER AGENT 1
# -----------------------------

print("\n[1] Peer Agent 1")

peer1 = run_agent(
    "Peer Agent 1",
    "Independent puzzle-solving peer.",
    """
Solve the puzzle independently.
Use direct logical reasoning.
Provide your proposed answer and explanation.
""",
    puzzle
)

print(peer1["response"])


# -----------------------------
# PEER AGENT 2
# -----------------------------

print("\n[2] Peer Agent 2")

peer2 = run_agent(
    "Peer Agent 2",
    "Independent puzzle-solving peer.",
    """
Solve the puzzle independently using a different reasoning approach.
Check all clues carefully.
Provide your proposed answer and explanation.
""",
    puzzle
)

print(peer2["response"])


# -----------------------------
# PEER AGENT 3
# -----------------------------

print("\n[3] Peer Agent 3")

peer3 = run_agent(
    "Peer Agent 3",
    "Independent verification peer.",
    """
Solve the puzzle independently.
Focus on checking whether possible answers satisfy every clue.
Give the most likely correct answer.
""",
    puzzle
)

print(peer3["response"])


# ============================================================
# PEER-TO-PEER INFORMATION EXCHANGE
# ============================================================

print("\n")
print("-" * 70)
print("PEER-TO-PEER INFORMATION EXCHANGE")
print("-" * 70)

peer_exchange_prompt = f"""
You are a peer-to-peer coordination agent.

Original Puzzle:
{puzzle}

Peer Agent 1:
{peer1["response"]}

Peer Agent 2:
{peer2["response"]}

Peer Agent 3:
{peer3["response"]}

Compare the three peer solutions.

Determine:
1. Which reasoning is consistent with the puzzle.
2. Where the peers agree.
3. Where they disagree.
4. The final solution based on logical verification.

Do not simply choose an answer by majority.
Check the actual puzzle constraints.

Final Answer:
"""

peer_exchange = ask_llm(
    peer_exchange_prompt,
    max_tokens=250
)

print(peer_exchange)


# ============================================================
# FINAL GLOBAL COORDINATOR
# ============================================================

final_prompt = f"""
You are the Global Coordinator.

Solve the puzzle using the results from three architectures.

PUZZLE:
{puzzle}

CENTRALIZED RESULT:
{central_result["response"]}

HIERARCHICAL RESULT:
{hierarchical_answer}

PEER-TO-PEER RESULT:
{peer_exchange}

Give the final response using exactly this format:

FINAL PUZZLE ANSWER:
Write only the correct answer.

REASONING:
Give a short explanation of how the answer was obtained.
Mention whether the three architectures agree or disagree.

ARCHITECTURE AGREEMENT:
State whether Centralized, Hierarchical, and Peer-to-Peer
architectures reached the same answer.

Do not repeat the instructions.
Do not add extra sections.
"""

final_answer = ask_llm(
    final_prompt,
    max_tokens=250
)

print("\n" + "=" * 70)
print("FINAL SOLUTION")
print("=" * 70)
print(final_answer)

# ============================================================
# ARCHITECTURE SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("MULTI-AGENT ARCHITECTURE SUMMARY")
print("=" * 70)

print("""
                         PUZZLE INPUT
                              |
             +----------------+----------------+
             |                |                |
             v                v                v
       CENTRALIZED       HIERARCHICAL     PEER-TO-PEER
             |                |                |
             |          +-----+-----+      +---+---+
             |          |     |     |      |   |   |
             |          v     v     v      v   v   v
             |        Lead Constraint Pattern Peer1 Peer2 Peer3
             |              |       |        \  |  /
             |              +-------+---------+---+
             |                      |
             |                Validator
             |                      |
             +----------------------+----------------+
                                    |
                                    v
                         GLOBAL COORDINATOR
                                    |
                                    v
                            FINAL SOLUTION
""")

print("=" * 70)
print("               PROJECT COMPLETED")
print("=" * 70)
```
